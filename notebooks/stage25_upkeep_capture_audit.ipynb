## Single-game exact-identity capture check

Runs one original game (first seed, candidate seat 0) **without** capture, then the same game
**with** capture under its original episode ID and sampling configuration. Compares outcome and
episode identity, then checks the capture's internal trace/rollout consistency and full status history.
**Stop and investigate on any mismatch; do not run the panel.**
If a previous results ZIP is available, its matching row can be compared against the no-capture row here
before proceeding.


```python
CHECK_SEED, CHECK_SEAT = SEEDS[0], 0
CHECK_FILTER = f"{CHECK_SEED}:{CHECK_SEAT}"
CHECK_EPISODE = MASTER_SEED * (2 * len(SEEDS)) + 2 * SEEDS.index(CHECK_SEED) + CHECK_SEAT
print('check game:', CHECK_FILTER, 'original episode:', CHECK_EPISODE, flush=True)

def run_eval(extra_args, log_path):
    command = [sys.executable, '-m', 'tools.evaluate_stage25_upkeep',
               '--checkpoint', str(PPO_CHECKPOINT), '--e-checkpoint', str(BC_E_CHECKPOINT),
               '--backend', 'official', '--e-history-version', 'E_LEGACY',
               '--master-seed', str(MASTER_SEED), '--seeds', *map(str, SEEDS),
               '--variants', 'baseline', '--game-filter', CHECK_FILTER, *extra_args]
    print('Starting:', ' '.join(command), flush=True)
    started = time.monotonic()
    with open(log_path, 'w') as log:
        process = subprocess.Popen(command, cwd=REPO, env=EVAL_ENV, text=True,
                                   stdout=subprocess.PIPE, stderr=subprocess.STDOUT, bufsize=1)
        try:
            for line in process.stdout:
                log.write(line)
                log.flush()
                print(line, end='', flush=True)
            returncode = process.wait()
        except BaseException:
            process.send_signal(signal.SIGINT)
            try:
                process.wait(timeout=10)
            except subprocess.TimeoutExpired:
                process.kill()
                process.wait()
            raise
    if returncode:
        raise RuntimeError(f'Evaluation exited {returncode}; inspect {log_path}.')
    print(f'Done in {(time.monotonic()-started)/60:.1f} minutes.', flush=True)

CHECK_NOCAP = RUN_ROOT / 'check_nocap'
CHECK_CAP_OUT = RUN_ROOT / 'check_cap'
CHECK_CAP = RUN_ROOT / 'check_captures'
run_eval(['--output-dir', str(CHECK_NOCAP)], RUN_ROOT / 'check_nocap.log')
run_eval(['--output-dir', str(CHECK_CAP_OUT), '--capture-dir', str(CHECK_CAP)],
         RUN_ROOT / 'check_cap.log')

row_off = [json.loads(line) for line in (CHECK_NOCAP / 'games.jsonl').read_text().splitlines()]
row_on = [json.loads(line) for line in (CHECK_CAP_OUT / 'games.jsonl').read_text().splitlines()]
assert len(row_off) == len(row_on) == 1, (row_off, row_on)
row_off, row_on = row_off[0], row_on[0]
for field in ('episode_id', 'seed', 'seat', 'bank', 'opponent_bank', 'statuses'):
    assert row_off[field] == row_on[field], (field, row_off[field], row_on[field])
assert row_on['episode_id'] == CHECK_EPISODE, (row_on['episode_id'], CHECK_EPISODE)
assert row_on.get('capture_complete') is True, row_on
assert row_off['statuses'] == ['DONE', 'DONE'], row_off
print('Outcome + episode identity match:', json.dumps(row_on), flush=True)

import gzip
cap_game = next(iter([p for p in CHECK_CAP.glob('baseline/*/meta.json')] ) ).parent
meta = json.loads((cap_game / 'meta.json').read_text())
assert meta['complete'] is True and meta['episode_id'] == CHECK_EPISODE, meta
assert meta.get('has_official_replay') is True, meta
trace = json.loads(gzip.open(cap_game / 'debug_trace.json.gz', 'rt').read())
rollout = json.loads(gzip.open(cap_game / 'rollout.json.gz', 'rt').read())
trace_actions = [t['joint_actions'] for t in trace['turns'] if 'joint_actions' in t]
roll_actions = [{'0': {'farmer': j['seat_0'].get('farmer'), 'hands': j['seat_0'].get('hands'),
                        'market': j['seat_0'].get('market')},
                 '1': {'farmer': j['seat_1'].get('farmer'), 'hands': j['seat_1'].get('hands'),
                        'market': j['seat_1'].get('market')}}
                for j in rollout['joint_actions']]
assert trace_actions == roll_actions, 'rollout/trace action mismatch'
history = json.loads((cap_game / 'status_history.json').read_text())
bad = [(i, s) for i, s in enumerate(history) if any(v not in ('ACTIVE', 'DONE') for v in s)]
assert not bad, bad
assert history[-1] == ['DONE', 'DONE'], history[-1]
print(f"Capture internally consistent: {len(trace_actions)} decision turns, "
      f"{len(history)} status entries, replay present.", flush=True)
print('Single-game capture check PASSED; proceed to the panel.', flush=True)
```

## Run the paired capture panel

This cell stays in the foreground and prints each completed game plus its capture path. First inference
may compile without printing a game result. Each game and capture is saved immediately.
Interrupting the cell stops its child process; partial results remain available.
To rerun, choose a new output directory rather than overwriting partial results.
The two source replay seeds make this a diagnostic screen, not promotion evidence.


```python
command = [sys.executable, '-m', 'tools.evaluate_stage25_upkeep',
           '--checkpoint', str(PPO_CHECKPOINT), '--e-checkpoint', str(BC_E_CHECKPOINT),
           '--output-dir', str(OUTPUT), '--capture-dir', str(CAPTURES), '--backend', 'official',
           '--e-history-version', 'E_LEGACY', '--master-seed', str(MASTER_SEED),
           '--seeds', *map(str, SEEDS), '--variants', *VARIANTS]
(RUN_ROOT / 'launch.json').write_text(json.dumps({'command':command, 'cwd':str(REPO),
    'source_commit':CODE_SHA, 'jax_platform':JAX_PLATFORM}, indent=2))
print('Starting', len(SEEDS) * 2 * len(VARIANTS), 'games; logs:', RUN_ROOT / 'run.log', flush=True)
started = time.monotonic()
with (RUN_ROOT / 'run.log').open('w') as log:
    process = subprocess.Popen(command, cwd=REPO, env=EVAL_ENV, text=True,
                               stdout=subprocess.PIPE, stderr=subprocess.STDOUT, bufsize=1)
    try:
        for line in process.stdout:
            log.write(line)
            log.flush()
            print(line, end='', flush=True)
        returncode = process.wait()
    except BaseException:
        process.send_signal(signal.SIGINT)
        try:
            process.wait(timeout=10)
        except subprocess.TimeoutExpired:
            process.kill()
            process.wait()
        raise
if returncode:
    raise RuntimeError(f'Evaluation exited {returncode}; inspect {RUN_ROOT / "run.log"}. Partial outputs preserved.')
print(f'Completed in {(time.monotonic()-started)/60:.1f} minutes.')
```

```python
audit_cmd = [sys.executable, '-m', 'tools.audit_stage25_capture',
             '--capture-dir', str(CAPTURES), '--output-dir', str(AUDIT), '--focus-pairs', '4']
print('Starting:', ' '.join(audit_cmd), flush=True)
audit_log = subprocess.run(audit_cmd, cwd=REPO, env=EVAL_ENV, text=True,
                           stdout=subprocess.PIPE, stderr=subprocess.STDOUT)
(RUN_ROOT / 'audit.log').write_text(audit_log.stdout)
print(audit_log.stdout, flush=True)
if audit_log.returncode:
    raise RuntimeError(f'Audit exited {audit_log.returncode}; inspect {RUN_ROOT / "audit.log"}.')
report = (AUDIT / 'audit.md').read_text()
print('\n'.join(report.splitlines()[:80]))
print('... (full report: audit/audit.md)')
```

```python
import pandas as pd
from IPython.display import display, FileLink
rows = [json.loads(line) for line in (OUTPUT / 'games.jsonl').read_text().splitlines()]
games = pd.DataFrame(rows)
if len(games) != len(SEEDS) * 2 * len(VARIANTS):
    raise RuntimeError('Incomplete panel; do not interpret as a finished comparison.')
if not all(status == ['DONE', 'DONE'] for status in games['statuses']):
    raise RuntimeError('Invalid game status in panel.')
if not games.get('capture_complete', pd.Series([False])).all():
    raise RuntimeError('Incomplete captures; inspect the capture warnings before auditing.')
base = games[games.variant == 'baseline'][['seed','seat','bank','opponent_bank']]
paired = games.merge(base, on=['seed','seat'], suffixes=('', '_baseline'), validate='many_to_one')
paired['bank_delta'] = paired.bank - paired.bank_baseline
paired['margin_delta'] = (paired.bank-paired.opponent_bank) - (paired.bank_baseline-paired.opponent_bank_baseline)
paired['win'] = (paired.bank > paired.opponent_bank).astype(int)
summary = paired.groupby('variant', sort=False).agg(
    games=('bank','size'), mean_bank=('bank','mean'), median_bank=('bank','median'),
    mean_bank_delta=('bank_delta','mean'), min_bank_delta=('bank_delta','min'),
    mean_margin_delta=('margin_delta','mean'), wins=('win','sum'))
display(summary.round(1))
display(paired[['variant','seed','seat','episode_id','bank','opponent_bank','bank_delta','margin_delta']])
summary.to_csv(OUTPUT / 'summary.csv')
paired.to_csv(OUTPUT / 'paired_games.csv', index=False)
# Bundle results + captures + audit; never checkpoints or secrets.
import zipfile
bundle_path = RUN_ROOT / 'stage25_capture_audit_results.zip'
with zipfile.ZipFile(bundle_path, 'w', zipfile.ZIP_DEFLATED, compresslevel=6) as bundle:
    for root in (OUTPUT, CAPTURES, AUDIT):
        for path in sorted(root.rglob('*')):
            if path.is_file():
                bundle.write(path, path.relative_to(RUN_ROOT))
    for extra in ('launch.json', 'run.log', 'audit.log'):
        bundle.write(RUN_ROOT / extra, extra)
display(FileLink(str(bundle_path)))
print('Send back the ZIP (or summary.csv, paired_games.csv, audit/audit.md + audit/*.csv).')
print('Four games per arm is a screen: confirm promising changes on more fresh seeds before adoption.')
print('The audit proposes evidence-linked bottleneck hypotheses only; no fix is implemented here.')
```
